```python
# 특성 변환(Feature Transformation)을 통해 직선으로 나눌 수 없던 데이터를
# 선형 분리(Linear Separation)가 가능하도록 바꾸는 과정 시각화
```

```python
import matplotlib.pyplot as plt

# 임의의 예제 데이터 생성: 각 점의 좌표는 (x1, x2)
# A: (-1, 1), (1, 1)
# B: (-3, 1), (3, 1)
x1 = [-1, 1, -3, 3]
x2 = [1, 1, 1, 1]
colors = ["blue", "blue", "red", "red"]
```

```python
x1
```

```python
x2
```

```python
# x1만 제곱
x1_squared = [x ** 2 for x in x1] # 리스트 컴프리헨션: 반복문을 한 줄로 줄여 결과를 바로 새 리스트로 묶어주는 파이썬 문법
# for x in x1 ... x1 리스트([-1, 1, -3, 3])에서 숫자를 하나씩 꺼내 x에 할당
# x ** 2: 꺼낸 숫자 x를 제곱
```

```python
x1_squared
```

```python
# 변환 전
plt.subplot(1, 2, 1)
plt.scatter(x1, x2, c=colors, s=100)
# x1, x2: 각 점의 가로(x축)와 세로(y축) 좌표.
  # 순서대로 짝지어 (-1, 1), (1, 1), (-3, 1), (3, 1) 위치에 plotting
# c=colors: 점의 색상(color) 지정.
  # 위에 지정한  ["blue", "blue", "red", "red"] 리스트에 맞춰 앞의 두 점은 파란색, 뒤의 두 점은 빨간색으로 표현
# s=100: 점의 크기(size) 지정

plt.title("Before")
plt.xlabel("x1")
plt.ylabel("x2")
plt.xlim(-4, 4)



# 변환 후
plt.subplot(1, 2, 2)
plt.scatter(x1_squared, x2, c=colors, s=100)
plt.axvline(x=5, color="black", linestyle="--")
plt.title("After: x1 squared")
plt.xlabel("x1 squared")
plt.xlim(0, 10)

plt.tight_layout()
plt.show()
```
